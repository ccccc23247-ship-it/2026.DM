<h3>그래프 실습</h3>

DSF 탐색

```python
graph = {'Alice':['Bob','Charlie'],'Bob':['David','Eve'],'Charlie':['Frank'],'David':[],'Eve':['Frank'],'Frank':[]}

def dfs(graph,v,visited):
  visited.add(v)
  print(v,end=' ')
  for neighber in graph[v]:
    if neighber not in visited:
      dfs(graph,neighber,visited)

print("DSF 탐색 순서")
dfs(graph,'Alice',set())
```

BFS 탐색

```python
from collections import deque

graph = {'Alice':['Bob','Charlie'],'Bob':['David','Eve'],'Charlie':['Frank'],'David':[],'Eve':['Frank'],'Frank':[]}

def bfs(graph, start):
  visited = set()
  queue = deque([start])
  visited.add(start)

  while queue:
    v = queue.popleft()
    print(v, end=' ')
    for neighbor in graph[v]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

print("\nBFS 탐색 순서")
bfs(graph, 'Alice')
```

<h3>친구 찾기</h3>

```python
from collections import deque

graph = {'Alice':['Bob','Charlie'],
         'Bob':['David','Eve'],
         'Charlie':['Frank'],
         'David':[],
         'Eve':['Frank'],
         'Frank':[]}

def recommend_friend(graph, start):
  visited = set()
  queue = deque([(start,0)])

  print(f"\n{start}친구 추천 2단계")

  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 2:
        print("-", current)
      if depth < 2:
        for neighber in graph[current]:
          queue.append((neighber,depth + 1))

recommend_friend(graph, 'Alice')
```
