1) 함수의 정의와 시각화

y = 2x + 1

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
  return 2*x + 1

x = np.linspace(-10,10,100)
y = f(x)

plt.plot(x,y,label='f(x) = 2x + 1')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Example of Function Gragh')
plt.legend()
plt.grid(True)
plt.show()
```
